## The paper

The paper presents qlPCA as an approach that recovers the spline-based algorithm of CATPCA [`meulman2004principal`] but **introduces continuous variables directly,
without a discretisation step**. Section V is a deliberately modest illustration — the authors say it is "a convenient illustration of our approach rather than a substantive
economic application":

* **Data:** twelve economic indicators for **26 European countries**, publicly available from Eurostat (Table I of the paper: trade balance in goods and in services, harmonised
  consumer prices, unemployment rate, long-term government bond yields, GDP per capita in PPS, real GDP growth, expenditure on pensions, balance of payments,
  motorisation rate, human resources in science and technology, public balance). The Eurostat codes are in `data/wce2011_eurostat_indicators.csv`.
* **Methods compared:** qlPCA (splines of degree 1 with one interior knot, no order restrictions), CATPCA (same splines, on variables discretised by the *multiplying*
  option) and ordinary PCA.
* **Explained variance in two dimensions (Table II):** qlPCA **58 %**, CATPCA **64 %**, PCA **54 %**. The authors note that CATPCA does better than qlPCA and that
  qlPCA is "not intended to be a direct competitor of CATPCA but rather a different approach".
* **Optimal transformations (Fig. 2):** although no order restriction was imposed, **all twelve transformations turn out monotonically increasing**. This is what allows the
  loadings to be read on the original variables.
* **Loadings (Table III)** of the two-dimensional qlPCA solution, and the **projection of the countries** on the first two nonlinear components (Fig. 3). The paper's
  interpretation: high positive scores on the first component go with high long-term bond yields and low GDP per capita; high positive scores on the second with high unemployment
  and low balance of payments.
